# Threads

```python
import time

def task_a():
    print("Starting task A")
    time.sleep(1)
    print("finish the task A")
    
def task_b():
    print("Starting task B")
    time.sleep(1)
    print("finish the task B")
    
start_time = time.perf_counter()
task_a()
task_b()
end_time = time.perf_counter()

print(f"Synchronous execution time: {end_time - start_time : .2f} seconds")
```

```python
import time
import threading

def task_a():
    print("Starting task A")
    time.sleep(1)
    print("finish the task A")
    
def task_b():
    print("Starting task B")
    time.sleep(1)
    print("finish the task B")
    
start_time2 = time.perf_counter()

thread1 = threading.Thread(target=task_a)
thread2 = threading.Thread(target=task_b)

thread1.start()
thread2.start()

thread1.join()
thread2.join()

end_time2 = time.perf_counter()

print(f"Threading execution time: {end_time2 - start_time2 : .2f} seconds")
```

```python
import threading

counter = 0

def increment():
    global counter
    for _ in range(100000):
        counter += 1

threads = []

for _ in range(10):
    t = threading.Thread(target=increment)
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print(f"Final counter value: {counter}")
```

```python
import threading
counter = 0
lock = threading.Lock()

def increment():
    global counter
    for _ in range(100000):
        lock.acquire()
        try:
            counter += 1
        finally:
            lock.release()

threads = []
for _ in range(10):
    t = threading.Thread(target=increment)
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print(f"Final counter value: {counter}")
```

```python
import time
import multiprocessing

def square_n(n):
    return n * n

if __name__ == "__main__":
    cpu_core = multiprocessing.cpu_count()
    print(f"No.of CPU cores: {cpu_core}")
    
    inputs = list(range(100_000_000))
    
    start_time = time.perf_counter()
    
    with multiprocessing.Pool(processes=cpu_core) as pool:
        result = pool.map(square_n, inputs)
    
    end_time = time.perf_counter()
    
    print(f"Pool execution took: {end_time - start_time: .2f} seconds.\n")
```

# IPC (Inter Process Communication)

```python
import time
import multiprocessing



def producer(q):
    for i in range(5):
        item = f"item {i}"
        q.put(item)
        print(f"Producer Put {item}")
        time.sleep(1)
    q.put(None)  # Signal the consumer to exit

def consumer(q):
    while True:
        item = q.get()
        if item is None:
            break
        print(f"Consumer Got {item}")

if __name__ == "__main__":
    queue = multiprocessing.Queue()
    p_producer = multiprocessing.Process(target=producer, args=(queue,))
    p_consumer = multiprocessing.Process(target=consumer, args=(queue,))

    p_producer.start()
    p_consumer.start()

    p_producer.join()
    p_consumer.join()

    print("Producer and Consumer have finished execution.")
```

# asyncio: Please use theses below code snippets in .py files individually to see the responses.

```python
import asyncio
import time


async def task_a():
    print("Starting task A")
    await asyncio.sleep(1)
    print("finish the task A")

async def task_b():
    print("Starting task B")
    await asyncio.sleep(1)
    print("finish the task B")

async def main():
    start_time = time.perf_counter()

    await asyncio.gather(
        task_a(),
        task_b()
    )

    end_time = time.perf_counter()
    print(f"Asyncio execution time: {end_time - start_time : .2f} seconds")

if __name__ == "__main__":
    asyncio.run(main())
```

```python
import asyncio
import time

lock = asyncio.Lock()

counter = 0

async def increment_safely():
    global counter
    async with lock:
        current = counter  
        await asyncio.sleep(0.01)  # Simulate some processing time
        counter = current + 1

async def main():
    tasks = [asyncio.create_task(increment_safely()) for _ in range(100)]
    await asyncio.gather(*tasks)
    print(f"Final counter value: {counter}")

if __name__ == "__main__":
    start_time = time.time()
    asyncio.run(main())
    end_time = time.time()
    print(f"Execution time: {end_time - start_time:.4f} seconds")
```
