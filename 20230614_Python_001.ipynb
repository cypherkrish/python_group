# Decorators

```python
def decor(func):
    def inner():
        value = func()
        return value + 2
    return inner
```

```python
def numn():
    return 5
```

```python
numn()
```

```python
result_num = decor(numn)
print(result_num())
```

```python
def decor(func):
    def inner():
        value = func()
        return value + 2
    return inner
```

```python
@decor
def numn():
    return 5
```

```python
print(numn())
```

```python
def decor1(func):
    def inner():
        value = func()
        return value + 2
    return inner

def decor2(func):
    def inner():
        value = func()
        return value * 2
    return inner

@decor2
@decor1
def num():
    return 10
print(num())
```

# Generators

```python
def generator_1(x, y):  # This is generator function
    while x <= y:  # Loop condition for yield
        yield x  # return value is x
        x += 1  # increment
```

```python
g = generator_1(1, 10)  # This will return generator object
```

```python
for i in g:
    print(i)  # This will print values from 1 to 10
```

```python
print(list(generator_1(1, 10)))
```

```python
def gen1():
    yield 'A'
    yield 'B'
    yield 'C'

g = gen1()

print(next(g))  # This will print 'A'
print(next(g))  # This will print 'B'
print(next(g))  # This will print 'C'
print(next(g))  # This will raise StopIteration error
```

```python
class MyIterator:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.end:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1
```

```python
my_iter = MyIterator(10, 15)
```

```python
for item in my_iter:
    print(item)  # This will print values from 10 to 15
```

```python
def square_generator():
    for x in range(1, 10):
        yield x ** 2
```

```python
squares = square_generator()
```

```python
for i in squares:
    print(i)  # This will print squares of numbers from 1 to 9
```
