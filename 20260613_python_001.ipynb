```python
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self):
        pass
    
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
```

```python
circle = Circle(5)
rectangle = Rectangle(4, 6)

print(f"Area of the circle: {circle.area()}")
print(f"Area of the rectangle: {rectangle.area()}")
```

```python
from abc import ABC, abstractmethod

class LibraryItem(ABC):
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.checked_out = False

    @abstractmethod
    def display_details(self):
        pass

    def check_out(self):
        if not self.checked_out:
            self.checked_out = True
            print(f"{self.title} by {self.author} has been checked out.")
        else:
            print(f"{self.title} by {self.author} is already checked out.")

    def check_in(self):
        if self.checked_out:
            self.checked_out = False
            print(f"{self.title} by {self.author} has been checked in.")
        else:
            print(f"{self.title} by {self.author} is already checked out.")
```

```python
class Book(LibraryItem):
    def __init__(self, title, author, genre):
        super().__init__(title, author)
        self.genre = genre

    def display_details(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Genre: {self.genre}")
        print(f"Checked Out: {'Yes' if self.checked_out else 'No'}")
```

```python
class DVD(LibraryItem):
    def __init__(self, title, director, runtime):
        super().__init__(title, director)
        self.duration = runtime

    def display_details(self):
        print(f"Title: {self.title}")
        print(f"Director: {self.author}")
        print(f"Duration: {self.duration} minutes")
        print(f"Checked Out: {'Yes' if self.checked_out else 'No'}")
```

```python
class Library:
    def __init__(self):
        self.catalog = []

    def add_item(self, item):
        self.catalog.append(item)
        print(f"{item.title} has been added to the library.")

    def display_catalog(self):
        for item in self.catalog:
            item.display_details()
            print()
```

```python
# Main program


if __name__ == "__main__":
    book1 = Book("The Great Gatsby", "F. Scott Fitzgerald", "Fiction")
    dvd1 = DVD("Inception", "Christopher Nolan", 148)
    book2 = Book("To Kill a Mockingbird", "Harper Lee", "Fiction")

    library = Library()

    library.add_item(book1)
    library.add_item(dvd1)
    library.add_item(book2)
    library.display_catalog()

    book1.check_out()
    dvd1.check_out()
    book1.check_out()
    library.display_catalog()
    
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"
```

```python
v1 = Vector(2, 3)
v2 = Vector(4, 5)
v3 = v1 + v2
print(v3)
v3 = v1 - v2
print(v3)
```

```python
class singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super(singleton, cls).__new__(cls)
        return cls._instance
```

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        return f"{self.name} says Woof!"

class Cat:
    def __init__(self, name):
        self.name = name

    def meow(self):
        return f"{self.name} says Meow!"
```

```python
def create_pet(pet):
    if pet == "dog":
        return Dog()
    elif pet == "cat":
        return Cat()
    else:
        raise ValueError("Unknown pet type")
```

# File Handeling

r : read (default)
w : write
a : append
b : binary
rb : read binary

```python
file = open("example.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
with open("example.txt", "r") as file:
    for line in file:
        print(line)
```

```python
with open("image1.png", "rb") as file:
    content = file.read()
    print(content)
```

```python
# Write to a file
with open("output.txt", "w") as file:
    file.write("Hello, World!")
    file.write("\nThis is a new line.")
```

# Exception handeling

# try, except, else and finally

```python
def divide (x, y):
    if y == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    return x / y
```

```python
try:
    result = divide(10, 0)

except ZeroDivisionError as e:
    print(e)
```

```python
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    result = x / y
    print(f"The result of {x} divided by {y} is: {result}")

except:
    print("An error occurred while performing the division.")
```

```python
try:
    x = int(input("Enter a number: "))
    y = int(input("Enter another number: "))
    print(x)
    print(y)
    result = x / y
    print(f"The result of {x} divided by {y} is: {result}")

except ZeroDivisionError:
    print("Cannot divide by zero. Please provide a non-zero divisor.")
except ValueError:
    print("Invalid input. Please enter valid numbers.")
```

```python
try:
    x = int(input("Enter a number: "))
    
except ValueError:
    print("Invalid input. Please enter valid numbers.")
else:
    print(f"You entered: {x}")
```

```python
try:
    x = int(input("Enter a number: "))
    
except Exception as e:
    print("Error occurred:", e)
else:
    print(f"You entered: {x}")
```

```python
try:
    x = int(input("Enter a number: "))
    if x < 0:
        raise ValueError("Negative number entered.")
    if x == 0:
        raise ZeroDivisionError("Zero is not allowed.")
except Exception as e:
    print("Error occurred:", e)
else:
    print(f"You entered: {x}")
finally:
    print("Execution completed.")
```
